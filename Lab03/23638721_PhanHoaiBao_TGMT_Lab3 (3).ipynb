```python
from PIL import Image
import os
from google.colab import drive
from IPython.display import display
```

```python
def bai3_xoay_lat_anh():
  print("\n=== BÀI 3: XOAY VÀ LẬT ẢNH ===")
  # Kết nối Google Drive
  drive.mount('/content/drive')
  try:
    # Mở ảnh - THAY ĐỔI ĐƯỜNG DẪN TẠI ĐÂY
    img = Image.open("/content/drive/MyDrive/bai2_resize_ratio.png")
    print(f"Kích thước ảnh gốc: {img.size}")
    # Hiển thị ảnh gốc
    print("\nẢNH GỐC:")
    display(img)
    # 1. Xoay 90, 180, 270 độ
    rotated_90 = img.rotate(90, expand=True)
    rotated_90.save("/content/drive/MyDrive/bai3_rotate_90.png")
    print("\nĐã xoay 90 độ")
    print("XOAY 90 ĐỘ:")
    display(rotated_90)
    rotated_180 = img.rotate(180)
    rotated_180.save("/content/drive/MyDrive/bai3_rotate_180.png")
    print("\nĐã xoay 180 độ")
    print("XOAY 180 ĐỘ:")
    display(rotated_180)
    rotated_270 = img.rotate(270, expand=True)
    rotated_270.save("/content/drive/MyDrive/bai3_rotate_270.png")
    print("\nĐã xoay 270 độ")
    print("XOAY 270 ĐỘ:")
    display(rotated_270)
    # 2. Lật ảnh
    flipped_horizontal = img.transpose(Image.FLIP_LEFT_RIGHT)
    flipped_horizontal.save("/content/drive/MyDrive/bai3_flip_horizontal.png")
    print("\nĐã lật theo chiều ngang")
    print("LẬT NGANG:")
    display(flipped_horizontal)
    flipped_vertical = img.transpose(Image.FLIP_TOP_BOTTOM)
    flipped_vertical.save("/content/drive/MyDrive/bai3_flip_vertical.png")
    print("\nĐã lật theo chiều dọc")
    print("LẬT DỌC:")
    display(flipped_vertical)
    # 3. Xoay góc tùy ý (45 độ)
    rotated_45 = img.rotate(45, expand=True, fillcolor='white')
    rotated_45.save("/content/drive/MyDrive/bai3_rotate_45.png")
    print("\nĐã xoay 45 độ")
    print("XOAY 45 ĐỘ:")
    display(rotated_45)
    # So sánh kích thước
    print(f"\nSO SÁNH KÍCH THƯỚC:")
    print(f"Ảnh gốc: {img.size}")
    print(f"Xoay 90°: {rotated_90.size}")
    print(f"Xoay 180°: {rotated_180.size}")
    print(f"Xoay 270°: {rotated_270.size}")
    print(f"Lật ngang: {flipped_horizontal.size}")
    print(f"Lật dọc: {flipped_vertical.size}")
    print(f"Xoay 45°: {rotated_45.size}")
  except Exception as e:
    print(f"Lỗi: {e}")
    # Chạy bài 3
```

```python
bai3_xoay_lat_anh()
```
