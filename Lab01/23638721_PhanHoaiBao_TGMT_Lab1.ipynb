# Bài 1: Làm quen với Pillow

```python
from PIL import Image
import os
from google.colab import drive
from IPython.display import display
```

```python
def bai1_lam_quen_pillow_drive():
  # Bài 1: Làm quen với Pillow - Sử dụng ảnh từ Google Drive
  print("=== BÀI 1: LÀM QUEN VỚI PILLOW (GOOGLE DRIVE) ===")
  # Mount Google Drive
  print("Đang kết nối với Google Drive...")
  drive.mount('/content/drive')
  print("Kết nối thành công!")
  # Liệt kê các file ảnh có sẵn trong Drive (tùy chọn)
  print("\nĐang tìm kiếm file ảnh trong thư mục gốc của Drive...")
  drive_path = "/content/drive/MyDrive"
  # Tìm file ảnh phổ biến
  image_extensions = ['.jpg', '.jpeg', '.png', '.bmp', '.gif', '.tiff']
  found_images = []
  try:
    for file in os.listdir(drive_path):
      if any(file.lower().endswith(ext) for ext in image_extensions):
        found_images.append(file)
    if found_images:
         print("Các file ảnh tìm thấy:")
         for i, img in enumerate(found_images[:5]):  # Chỉ hiển thị 5 file đầu
            print(f" {i+1}. {img}")
         if len(found_images) > 5:
            print(f" ... và {len(found_images) - 5} file khác")
         else:
            print("Không tìm thấy file ảnh nào trong thư mục gốc")
    else:
        print("Không tìm thấy file ảnh nào trong thư mục gốc")
  except:
      print("Không thể truy cập thư mục Drive")
  # Đường dẫn ảnh - THAY ĐỔI TÊN FILE Ở ĐÂY
  image_filename = "image7.jpg"  # <=== THAY ĐỔI TÊN FILE TẠI ĐÂY
  image_path = f"/content/drive/MyDrive/Mia_Seb.png"
  print(f"\nĐang thử mở ảnh: {image_path}")
  try:
  # Kiểm tra file có tồn tại không
    if not os.path.exists(image_path):
      print(f"File không tồn tại: {image_path}")
      print("\nHƯỚNG DẪN SỬA LỖI:")
      print("1. Kiểm tra tên file trong biến 'image_filename'")
      print("2. Đảm bảo file ảnh đã được upload lên Google Drive")
      print("3. File phải nằm trong thư mục img/")
      print("4. Hoặc thay đổi đường dẫn đầy đủ nếu file ở thư mục khác")
      return
  # Mở ảnh
    img = Image.open(image_path)
    # In thông tin ảnh
    print(f"Mở ảnh thành công!")
    print(f"Đường dẫn: {image_path}")
    print(f"Kích thước: {img.size}")
    print(f"Chiều rộng: {img.width}")
    print(f"Chiều cao: {img.height}")
    print(f"Định dạng: {img.format}")
    print(f"Chế độ màu: {img.mode}")
    # Hiển thị ảnh
    print(f"\nHiển thị ảnh:")
    display(img)
    # Lưu ảnh với tên khác
    output_path = "/content/bai1_copy_drive.png"
    img.save(output_path)
    print(f"Đã lưu ảnh copy tại: {output_path}")
    # Lưu ảnh trở lại Drive (tùy chọn)
    output_drive_path = "/content/drive/MyDrive/bai1_copy_drive.png"
    img.save(output_drive_path)
    print(f"Đã lưu ảnh copy vào Drive: {output_drive_path}")
  except FileNotFoundError:
    print(f"Không tìm thấy file: {image_path}")
    print("\nHƯỚNG DẪN:")
    print("1. Upload ảnh lên Google Drive")
    print("2. Thay đổi tên file trong biến 'image_filename'")
    print("3. Hoặc sử dụng đường dẫn đầy đủ đến file")
  except Exception as e:
    print(f"Lỗi: {e}")
```

```python
print("  BẮT ĐẦU CHƯƠNG TRÌNH")
print("=" * 50)
bai1_lam_quen_pillow_drive()
```
